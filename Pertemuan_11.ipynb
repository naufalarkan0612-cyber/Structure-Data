```python
def coin_change_greedy(amount, coins):
    coins.sort(reverse=True)
    result = []
    for coin in coins:
        while amount >= coin:
            amount -= coin
            result.append(coin)
    return result

# Contoh penggunaan
amount = 68
coins = [3.14, 9.11, 19.45]
change = coin_change_greedy(amount, coins)
print("Koin yang digunakan:", change)

# menghitung hasil sebenarnya
result = 0
for i in change:
  result += i
print("Hasil yang sebenarnya:", result)
```

```python
def fractional_knapsack(items, capacity):
    # items = list of tuples (value, weight)
    items = sorted(items, key=lambda x: x[0]/x[1], reverse=True)
    total_value = 0.0
    tas = []
    for value, weight in items:
        if capacity >= weight:
            capacity -= weight
            tas.append((value, weight))
            total_value += value
        else:
            sisa = value * (capacity / weight)
            total_value += sisa
            tas.append((sisa, capacity))
            break
    return total_value, tas

# Contoh penggunaan
items = [(60, 10), (100, 20), (120, 30)]  # (nilai, berat)
capacity = 50
max_value, isi_tas = fractional_knapsack(items, capacity)
print("Nilai maksimum yang dapat dibawa:", max_value)
print("Isi tas:", isi_tas)
```

```python
import heapq

def prim_mst(graph, start):
    visited = set()
    min_heap = [(0, start, None)]
    total_weight = 0
    route = []

    while min_heap:
        weight, node, parent = heapq.heappop(min_heap)
        if node in visited:
            continue

        visited.add(node)
        total_weight += weight

        if parent is not None:
            route.append((parent, node, weight))
        for neighbor, edge_weight in graph[node]:
            if neighbor not in visited:
                heapq.heappush(min_heap, (edge_weight, neighbor, node))
    return total_weight, route

# Representasi graf: adjacency list
graph = {
    'A': [('B', 2), ('C', 3)],
    'B': [('A', 2), ('C', 1), ('D', 1)],
    'C': [('A', 3), ('B', 1), ('D', 4)],
    'D': [('B', 1), ('C', 4)]
}

jarak, rute = prim_mst(graph, 'A')
print(f"Total bobot MST (Prim): {jarak}")
print(f"Jalur yang dilewati: {rute}")
```

```python
def kruskal_mst(edges, n_nodes):
    parent = {i: i for i in range(n_nodes)}

    def find(x):
        while parent[x] != x:
            x = parent[x]
        return x

    def union(x, y):
        root_x = find(x)
        root_y = find(y)
        if root_x != root_y:
            parent[root_y] = root_x
            return True
        return False

    edges.sort(key=lambda x: x[2])
    total_weight = 0
    for u, v, weight in edges:
        if union(u, v):
            total_weight += weight
    return total_weight

# # edges: (node1, node2, weight), nodes are represented by integers
edges = [(0, 1, 2), (0, 2, 3), (1, 2, 1), (1, 3, 1), (2, 3, 4)]
print("Total bobot MST (Kruskal):", kruskal_mst(edges, 4))
```
