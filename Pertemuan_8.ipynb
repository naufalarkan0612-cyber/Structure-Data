Undirected Graph

```python
# Membuat graf tak berarah menggunakan adjacency list
graph = {
  1: [2, 3],
  2: [1, 3, 4, 5],
  3: [1, 2, 4],
  4: [2, 3, 5],
  5: [2, 4]
}

# Menampilkan graf
def print_graph(graf):
  for node in graf:
    print (f" {node} -> {graf [node]}")

print_graph(graph)
```

BFS dan DFS

```python
# BFS
from collections import deque

def bfs (graph, start):
  visited = set()
  queue = deque([start])

  while queue:
    vertex = queue.popleft ()
    if vertex not in visited:
      print(vertex, end=" ")
      visited.add (vertex)
      queue.extend([neighbor for neighbor in graph [vertex] if neighbor not in visited])

# Contoh pemanggilan
print("BFS:")
bfs(graph, 1)

# DFS
def dfs(graph, start, visited=None):
  if visited is None:
    visited = set()

  if start not in visited:
    print(start, end=" ")
    visited.add(start)
    for neighbor in graph[start]:
      dfs(graph, neighbor, visited)

# Contoh pemanggilan
print("\n\nDFS:")
dfs(graph, 1)
```

Return to BFS

```python
from collections import deque

def bfs (graph, start):
  visited = set()
  queue = deque([start])

  visit = []

  while queue:
    vertex = queue.popleft ()
    if vertex not in visited:
      visit.append(vertex)
      visited.add (vertex)
      queue.extend([neighbor for neighbor in graph [vertex] if neighbor not in visited])
  return visit

# Contoh pemanggilan
print("BFS:")
bfs(graph, 1)
```

Finding DFS

```python
def find_path(graph, start, end, visited=None):
  if visited is None:
    visited = set()
  visited.add(start)

  if start == end:
    return[start]

  for neighbor in graph[start]:
    if neighbor not in visited:
      path = find_path(graph, neighbor, end, visited)
      if path is not None:
        return[start] + path
  return None

print("Path finded:")
path = find_path(graph, 1, 4)
print(path)
```

We are DFS

```python
def is_connected(graph):
  if not graph:
    return False

  def dfs(node, visited):
    if node not in visited:
      visited.add(node)
      for neighbor in graph[node]:
        dfs(neighbor, visited)
  start_node = list(graph.keys())[0]
  visited_nodes = set()
  dfs(start_node, visited_nodes)
  return len(visited_nodes) == len(graph)

print("Is Graphs is connected to each other?")
print(is_connected(graph))
```
