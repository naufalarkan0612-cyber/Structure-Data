Soal 1

```python
class Node:
  def __init__(self, data):
    self.data = data
    self.left = None
    self.right = None

# Membuat tree secara manual
root = Node(19) # Naufal Arkan El Rochim.
root.left = Node(7) # Naufal Arkan El Rochim.
root.right = Node(68) # 2505060068
root.left.left = Node(24) # 24
root.right.right = Node(3) # Maret

# Fungsi inorder traversal
def inorder(node):
  if node:
    inorder(node.left)
    print(node.data, end=' ')
    inorder(node.right)

print("Hasil inorder traversal dari binary tree:")
print("Struktur Tree berdasarkan identitas saya.")
inorder(root)
```

Soal 2

```python
# Fungsi tree
class Node:
  def __init__(self, data):
    self.data = data
    self.left = None
    self.right = None

# Mohon maaf Pak, ini yang root kan di soal disuruh 3 digit terakhir, nah disitu angka depannya 0.
# Jadi, biar jalan harus saya jadikan string (semua, biar adil).

# Daftar angka untuk Node
array = ['068', '24', '3', '78', '97', '11', '7']

root = Node(array[0])
root.left = Node(array[1])
root.right = Node(array[2])
root.left.left = Node(array[3])
root.left.right = Node(array[4])
root.right.left = Node(array[5])
root.right.right = Node(array[6])

# inorder (left → root → right)
def inorder(node):
  if node:
    inorder(node.left)
    print(node.data, end=' ')
    inorder(node.right)

# preorder(root → left → right)
def preorder(node):
  if node:
    print(node.data, end=' ')
    preorder(node.left)
    preorder(node.right)

# postorder (left → right → root).
def postorder(node):
  if node:
    postorder(node.left)
    postorder(node.right)
    print(node.data, end=' ')

# Hasil inorder traversal dari binary tree.
print("Inorder Traversal:")
inorder(root)
# Hasil preorder traversal dari binary tree.
print("\nPreorder Traversal:")
preorder(root)
# Hasil postorder traversal dari binary tree.
print("\nPostorder Traversal:")
postorder(root)

# Mencari angka dengan dua digit terakhir NPM (tapi agak dicurangi)
# Karena tadi 3 digit dan sekarang 2 digit, mak a perlu bantuan (+ 0)
count = 0
print("\n\nApakah 2 Digit NPM terakhir saya (68) ada? ")
for i in array:
  if i == "0" + "69":
    print("Ketemu :)")
    break
  else:
    count += 1
  if count == len(array):
      print("Nggak ketemu :)")
      break

# Mencari angka yang tidak termasuk dalam list.
print("\n")
print(array)

inp = input("Mau cari angka apa? ")
for i in array:
  if i == inp:
    print("Ada :)")
  else:
    print(f"{inp} Nggak ada :(")
    break
```
