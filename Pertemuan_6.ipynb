Quick Sort

```python
# Quick Sortnya
def film_terlaris(arr):
  if len(arr) <= 1:
    return arr

  pivot_item = arr[len(arr) // 2]
  pivot_value = pivot_item[1]

  left = [x for x in arr if x[1] > pivot_value]
  middle = [x for x in arr if x[1] == pivot_value]
  right = [x for x in arr if x[1] < pivot_value]

  return film_terlaris(left) + middle + film_terlaris(right)

# Databasenya
data_film = {
    "Rainbow Ruby" : 544,
    "Dora the Explorer" : 388,
    "Sofia the First" : 583,
    "My Little Pony" : 944,
    "Tinker Bell" : 838,
    "Masha and the Bear" : 387,
    "Chibi Maruko-chan" :517
    }

# Contoh penggunaan
data = [[key, value] for key, value in data_film.items()]

sorted_data = film_terlaris(data)

print("Data sebelum sorting:")
for film in data:
  print(f"{film[0]}: {film[1]} penonton")

print("\nData setelah sorting:")
for film in sorted_data:
    print(f"{film[0]}: {film[1]} penonton")
```
